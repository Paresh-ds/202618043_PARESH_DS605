Task 2 - Vectorized Arithmetic and Linear Algebra

```python
import numpy as np

A = np.array([
    [1, 2],
    [4, 4]
])

B = np.array([
    [8, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
addition = A + B

print("Matrix Addition:")
print(addition)
element_multiplication = A * B

print("Element-wise Multiplication:")
print(element_multiplication)
matrix_multiplication = A @ B

print("Matrix Multiplication:")
print(matrix_multiplication)
```

```python
# Transpose
A_transpose = A.T

print("Transpose of A:")
print(A_transpose)

# Determinant
det_A = np.linalg.det(A)

print("\nDeterminant of A:", det_A)
```

```python
# Find inverse
A_inverse = np.linalg.inv(A)

print("Inverse of A:")
print(A_inverse)

# Verify A × A_inverse
verification = A @ A_inverse

print("\nA @ A_inverse:")
print(verification)

# Check whether result is approximately identity matrix
print("\nVerification using np.allclose():")
print(np.allclose(verification, np.eye(2)))

is_identity = np.allclose(verification, np.eye(2))

print("\nIs A @ A_inverse approximately identity?")
print(is_identity)
```

Task 3 - Normal Distribution and Histogram
● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard
deviation.
● Find the sample mean and sample standard deviation. Compare them with the chosen values.
● Plot a histogram with it.

```python
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Chosen mean and standard deviation
mean = 50
std = 10

# Generate 1000 values from normal distribution
data = np.random.normal(loc=mean, scale=std, size=1000)

print("Number of values:", len(data))
```

```python
sample_mean = np.mean(data)

print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)

sample_std = np.std(data, ddof=1)

print("Chosen Standard Deviation:", std)
print("Sample Standard Deviation:", sample_std)
```

```python
mean_difference = sample_mean - mean
std_difference = sample_std - std

print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)
```

```python
plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")

plt.show()
```

Task 6 - groupby() and Aggregation
● Survival rate by Sex.
● Survival rate by Pclass.
● Average Age and Fare by Pclass.
● Passenger count and survival rate by Sex-Pclass.
● Passenger count, average Fare, and survival rate by Embarked.

```python

# 1. Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("1. Survival Rate by Sex:")
print(survival_by_sex)


# 2. Survival rate by Pclass
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival Rate by Pclass:")
print(survival_by_class)


# 3. Average Age and Fare by Pclass
average_by_class = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(average_by_class)


# 4. Passenger count and survival rate by Sex-Pclass
sex_class_summary = df.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_class_summary["Survival_Rate"] *= 100

print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_class_summary)


# 5. Passenger count, average Fare and survival rate by Embarked
embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_summary["Survival_Rate"] *= 100

print("\n5. Passenger Count, Average Fare and Survival Rate by Embarked:")
print(embarked_summary)
```

Task 7 - Missing Values and Fare Outliers
● Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
● Fill missing Age values with the mean Age. Show missing counts before and after.
● Try to fill NaN values by imputing using: mean, median, mode, random value
● For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python


# Reload original dataset
df = pd.read_csv("train.csv")


# 1. Missing-value count and percentage

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

missing_summary["Missing_Percentage"] = (
    missing_summary["Missing_Percentage"].round(2)
)

print("Missing Value Summary:")
print(missing_summary)


# 2. Plot missing counts

missing_count = df.isnull().sum()
missing_count = missing_count[missing_count > 0]

plt.figure(figsize=(8, 5))
plt.bar(missing_count.index, missing_count.values)

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.title("Missing Values by Column")
plt.xticks(rotation=45)

plt.show()


# 3. Mean imputation

print("\nMissing Age before:", df["Age"].isnull().sum())

df_mean = df.copy()

mean_age = df_mean["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

print("Mean Age:", mean_age)
print("Missing Age after mean imputation:",
      df_mean["Age"].isnull().sum())


# 4. Median imputation

df_median = df.copy()

median_age = df_median["Age"].median()

df_median["Age"] = df_median["Age"].fillna(median_age)

print("\nMedian Age:", median_age)
print("Missing Age after median imputation:",
      df_median["Age"].isnull().sum())


# 5. Mode imputation

df_mode = df.copy()

mode_age = df_mode["Age"].mode()[0]

df_mode["Age"] = df_mode["Age"].fillna(mode_age)

print("\nMode Age:", mode_age)
print("Missing Age after mode imputation:",
      df_mode["Age"].isnull().sum())


# 6. Random-value imputation

df_random = df.copy()

np.random.seed(42)

missing_age = df_random["Age"].isnull()

random_values = np.random.choice(
    df_random.loc[~missing_age, "Age"],
    size=missing_age.sum(),
    replace=True
)

df_random.loc[missing_age, "Age"] = random_values

print("\nMissing Age after random imputation:",
      df_random["Age"].isnull().sum())


# 7. Fare outliers using IQR

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)


# Count outliers
outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Number of Fare outliers:", len(outliers))
```

Task 8 - Features and Pivot Table
● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest
groups.


```python

# 1. Create FamilySize

df["FamilySize"] = (
    df["SibSp"] +
    df["Parch"] +
    1
)

print("FamilySize:")
print(df[["SibSp", "Parch", "FamilySize"]].head())


# 2. Create IsAlone

df["IsAlone"] = np.where(
    df["FamilySize"] == 1,
    1,
    0
)

print("\nFamilySize and IsAlone:")
print(df[["FamilySize", "IsAlone"]].head(10))


# 3. Create Pivot Table

pivot = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("\nSurvival Rate Pivot Table:")
print(pivot)


# 4. Convert to percentage

pivot_percentage = pivot * 100

print("\nSurvival Rate (%):")
print(pivot_percentage.round(2))


# 5. Highest survival group

highest_group = pivot.stack().idxmax()
highest_rate = pivot.stack().max()

print("\nHighest Survival Group:")
print(highest_group)
print("Survival Rate:", round(highest_rate * 100, 2), "%")


# 6. Lowest survival group

lowest_group = pivot.stack().idxmin()
lowest_rate = pivot.stack().min()

print("\nLowest Survival Group:")
print(lowest_group)
print("Survival Rate:", round(lowest_rate * 100, 2), "%")
```

Task 9 - Visualizations and Observations
● Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative
relationships.
● Plot the survival rate by Sex and report the male/female survival rates.
● Plot Age vs Fare and distinguish Survived=0 and Survived=1.
● Write 5-7 short observations supported by your numerical results or plots.

```python

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["IsAlone"] = np.where(
    df["FamilySize"] == 1,
    1,
    0
)


# 1. Correlation Heatmap

numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation.round(2))


plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Dataset")
plt.tight_layout()
plt.show()


corr_matrix = correlation.copy()

corr_values = corr_matrix.to_numpy().copy()

np.fill_diagonal(corr_values, np.nan)

corr_matrix = pd.DataFrame(
    corr_values,
    index=correlation.index,
    columns=correlation.columns
)

positive_pair = corr_matrix.stack().idxmax()
positive_value = corr_matrix.stack().max()

negative_pair = corr_matrix.stack().idxmin()
negative_value = corr_matrix.stack().min()

print("\n------------------------------------------")
print("Strongest Relationships")
print("------------------------------------------")

print(
    "Strongest positive relationship:",
    positive_pair,
    "=",
    round(positive_value, 3)
)

print(
    "Strongest negative relationship:",
    negative_pair,
    "=",
    round(negative_value, 3)
)


# 2. Survival Rate by Sex

survival_by_sex = (
    df.groupby("Sex")["Survived"].mean() * 100
)

print("\n------------------------------------------")
print("Survival Rate by Sex")
print("------------------------------------------")

print(survival_by_sex.round(2))


plt.figure(figsize=(7, 5))

plt.bar(
    survival_by_sex.index,
    survival_by_sex.values
)

plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.title("Titanic Survival Rate by Sex")

plt.ylim(0, 100)

plt.tight_layout()
plt.show()


# 3. Age vs Fare by Survival Status

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    alpha=0.7
)

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare by Survival Status")

plt.tight_layout()
plt.show()



print("\n==========================================")
print("Additional Results for Observations")
print("==========================================")


# Overall survival rate
overall_survival = df["Survived"].mean() * 100

print(
    "\nOverall Survival Rate:",
    round(overall_survival, 2),
    "%"
)


# Survival rate by Pclass
survival_by_class = (
    df.groupby("Pclass")["Survived"].mean() * 100
)

print("\nSurvival Rate by Pclass:")
print(survival_by_class.round(2))


# Survival rate by Sex and Pclass
sex_class_survival = (
    df.groupby(["Sex", "Pclass"])["Survived"]
    .mean() * 100
)

print("\nSurvival Rate by Sex and Pclass:")
print(sex_class_survival.round(2))


# Missing values
print("\nMissing Values:")
print(df.isnull().sum())


# Fare outliers using IQR
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(fare_outliers))
```
