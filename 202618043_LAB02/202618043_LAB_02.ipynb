Task 2 - Vectorized Arithmetic and Linear Algebra

```python
import numpy as np

A = np.array([
    [1, 2],
    [4, 4]
])

B = np.array([
    [8, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
addition = A + B

print("Matrix Addition:")
print(addition)
element_multiplication = A * B

print("Element-wise Multiplication:")
print(element_multiplication)
matrix_multiplication = A @ B

print("Matrix Multiplication:")
print(matrix_multiplication)
```

```python
# Transpose
A_transpose = A.T

print("Transpose of A:")
print(A_transpose)

# Determinant
det_A = np.linalg.det(A)

print("\nDeterminant of A:", det_A)
```

```python
# Find inverse
A_inverse = np.linalg.inv(A)

print("Inverse of A:")
print(A_inverse)

# Verify A × A_inverse
verification = A @ A_inverse

print("\nA @ A_inverse:")
print(verification)

# Check whether result is approximately identity matrix
print("\nVerification using np.allclose():")
print(np.allclose(verification, np.eye(2)))

is_identity = np.allclose(verification, np.eye(2))

print("\nIs A @ A_inverse approximately identity?")
print(is_identity)
```

Task 3 - Normal Distribution and Histogram
● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard
deviation.
● Find the sample mean and sample standard deviation. Compare them with the chosen values.
● Plot a histogram with it.

```python
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Chosen mean and standard deviation
mean = 50
std = 10

# Generate 1000 values from normal distribution
data = np.random.normal(loc=mean, scale=std, size=1000)

print("Number of values:", len(data))
```

```python
sample_mean = np.mean(data)

print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)

sample_std = np.std(data, ddof=1)

print("Chosen Standard Deviation:", std)
print("Sample Standard Deviation:", sample_std)
```

```python
mean_difference = sample_mean - mean
std_difference = sample_std - std

print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)
```

```python
plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")

plt.show()
```

Task 5 - Filtering and Querying
Use Boolean indexing and/or query() to answer:
● Male passengers older than 50: how many?
● Female first-class passengers: how many, and what percentage survived?
● Age 20-40, Fare above overall median, and survived: how many?
● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("Male passengers older than 50:", len(male_over_50))
```

```python
print(type(male_over_50))
```

```python
female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

print("Female first-class passengers:", len(female_first_class))
#print(female_first_class.head(5))
#print(female_first_class.info())
survival_rate = female_first_class["Survived"].mean() * 100
print(survival_rate)
```

```python
fare_median = df["Fare"].median()

print("Overall median Fare:", fare_median)
result = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("Number of passengers:", len(result))
```

```python
alone_young_not_survived = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print(
    "Passengers travelling alone, age < 30, and did not survive:",
    len(alone_young_not_survived)
)
```

```python
southampton = df[df["Embarked"] == "S"]

southampton_median = southampton["Fare"].median()

print("Southampton median Fare:", southampton_median)
result = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("Number of passengers:", len(result))
```
