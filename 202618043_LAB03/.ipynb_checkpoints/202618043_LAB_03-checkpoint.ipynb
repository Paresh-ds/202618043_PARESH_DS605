```python
# Selected numerical columns for outlier analysis
outlier_columns = [
    "lead_time",
    "adr",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies"
]

print(outlier_columns)
```

```python
import matplotlib.pyplot as plt
```

```python
# Create boxplots for selected numerical features

for column in outlier_columns:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[column].dropna())
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.ylabel("Value")
    plt.show()
```

```python
# Detect outliers using the IQR method

outlier_summary = []

for column in outlier_columns:
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_table = pd.DataFrame(outlier_summary)

display(outlier_table)
```

Task 3 - Create Two Preprocessing Pipelines
● Split the data using train_test_split(test_size=0.2, stratify=y, random_state=42). Use the same split for all experiments.
● For numerical columns, use KNNImputer(n_neighbors=5) for missing values.
● For categorical columns, use SimpleImputer(strategy="most_frequent") and OneHotEncoder(handle_unknown="ignore").
● Pipeline A: KNNImputer + StandardScaler for numerical features.
● Pipeline B: KNNImputer + MinMaxScaler for numerical features.
● Use ColumnTransformer and Pipeline so that preprocessing is fitted only on the training data.

```python
from sklearn.model_selection import train_test_split

# Create target variable
y = df_clean["is_canceled"]

# Create input features
X = df_clean.drop(columns=["is_canceled"])

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Identify numerical columns
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Identify categorical columns
categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

# Numerical preprocessing
numerical_imputer = KNNImputer(n_neighbors=5)

# Categorical preprocessing
categorical_imputer = SimpleImputer(strategy="most_frequent")

categorical_encoder = OneHotEncoder(
    handle_unknown="ignore"
)

print("Numerical columns:", numerical_columns)
print("Categorical columns:", categorical_columns)
print("KNN Imputer created with 5 neighbors.")
print("Categorical imputer and encoder created.")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Numerical preprocessing for Pipeline A
numerical_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine numerical and categorical preprocessing
preprocessor_A = ColumnTransformer([
    ("num", numerical_pipeline_A, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])

# Create Pipeline A
pipeline_A = Pipeline([
    ("preprocessor", preprocessor_A)
])

print("Pipeline A created successfully.")
print(pipeline_A)
```

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Numerical preprocessing for Pipeline B
numerical_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# Combine numerical and categorical preprocessing
preprocessor_B = ColumnTransformer([
    ("num", numerical_pipeline_B, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns)
])

# Create Pipeline B
pipeline_B = Pipeline([
    ("preprocessor", preprocessor_B)
])

print("Pipeline B created successfully!")
print(pipeline_B)
```

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Numerical transformation
numerical_transformer = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

# Categorical transformation
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both transformations
preprocessor = ColumnTransformer([
    ("num", numerical_transformer, numerical_columns),
    ("cat", categorical_transformer, categorical_columns)
])

# Fit transformation ONLY on training data
X_train_transformed = preprocessor.fit_transform(X_train)

# Apply the already-fitted transformation to test data
X_test_transformed = preprocessor.transform(X_test)

print("Transformation completed successfully!")
print("Transformed X_train shape:", X_train_transformed.shape)
print("Transformed X_test shape:", X_test_transformed.shape)
```

Task 5 - Evaluate and Compare the Four Results
● For each experiment, calculate training accuracy, testing accuracy, precision, recall, and F1-score.
● Create a final comparison table containing all four model-pipeline combinations.
● Plot a confusion matrix for the best Logistic Regression result and the best Decision Tree result.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

# 1. Logistic Regression + Pipeline A
logistic_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])
logistic_A.fit(X_train, y_train)


# 2. Logistic Regression + Pipeline B
logistic_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])
logistic_B.fit(X_train, y_train)


# 3. Decision Tree + Pipeline A
decision_tree_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", DecisionTreeClassifier(random_state=42))
])
decision_tree_A.fit(X_train, y_train)


# 4. Decision Tree + Pipeline B
decision_tree_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", DecisionTreeClassifier(random_state=42))
])
decision_tree_B.fit(X_train, y_train)


print("All four models have been trained successfully!")
```

```python
from sklearn.metrics import accuracy_score

# Store all four trained models
models = {
    "Logistic Regression + Pipeline A": logistic_A,
    "Logistic Regression + Pipeline B": logistic_B,
    "Decision Tree + Pipeline A": decision_tree_A,
    "Decision Tree + Pipeline B": decision_tree_B
}

# Calculate training and testing accuracy
for name, model in models.items():

    # Predict training data
    y_train_pred = model.predict(X_train)

    # Predict testing data
    y_test_pred = model.predict(X_test)

    # Calculate accuracy
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)

    print(name)
    print("Training Accuracy:", train_accuracy)
    print("Testing Accuracy :", test_accuracy)
    print("-" * 50)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Store the four trained models
models = {
    "Logistic Regression + Pipeline A": logistic_A,
    "Logistic Regression + Pipeline B": logistic_B,
    "Decision Tree + Pipeline A": decision_tree_A,
    "Decision Tree + Pipeline B": decision_tree_B
}

# Calculate Precision, Recall and F1-score
for name, model in models.items():

    # Make predictions on test data
    y_test_pred = model.predict(X_test)

    # Calculate evaluation metrics
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    print(name)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1-score :", f1)
    print("-" * 50)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import pandas as pd

# Store all four trained models
models = {
    "Logistic Regression + Pipeline A": logistic_A,
    "Logistic Regression + Pipeline B": logistic_B,
    "Decision Tree + Pipeline A": decision_tree_A,
    "Decision Tree + Pipeline B": decision_tree_B
}

# Store the results
results = []

# Evaluate each model
for name, model in models.items():

    # Predictions on training data
    y_train_pred = model.predict(X_train)

    # Predictions on testing data
    y_test_pred = model.predict(X_test)

    # Calculate metrics
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    # Add results to the list
    results.append({
        "Model + Pipeline": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1
    })

# Convert results into a DataFrame
comparison_table = pd.DataFrame(results)

# Display the final comparison table
display(comparison_table)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# -----------------------------------------
# Best Logistic Regression - Pipeline A
# -----------------------------------------

y_pred_logistic = logistic_A.predict(X_test)

cm_logistic = confusion_matrix(y_test, y_pred_logistic)

disp_logistic = ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled", "Canceled"]
)

disp_logistic.plot()
plt.title("Confusion Matrix - Logistic Regression + Pipeline A")
plt.show()


# -----------------------------------------
# Best Decision Tree - Pipeline A
# -----------------------------------------

y_pred_tree = decision_tree_A.predict(X_test)

cm_tree = confusion_matrix(y_test, y_pred_tree)

disp_tree = ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled", "Canceled"]
)

disp_tree.plot()
plt.title("Confusion Matrix - Decision Tree + Pipeline A")
plt.show()
```

```python
# Calculate the difference between training and testing accuracy

for name, model in models.items():

    # Training prediction
    y_train_pred = model.predict(X_train)

    # Testing prediction
    y_test_pred = model.predict(X_test)

    # Calculate accuracies
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)

    # Calculate the performance gap
    performance_gap = train_accuracy - test_accuracy

    print(name)
    print("Training Accuracy :", round(train_accuracy, 4))
    print("Testing Accuracy  :", round(test_accuracy, 4))
    print("Performance Gap   :", round(performance_gap, 4))
    print("-" * 50)
```
